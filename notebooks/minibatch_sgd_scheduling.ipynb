```python
import numpy as np
import matplotlib.pyplot as plt

from src.objectives.quadratic import QuadraticObjective
# Import your original plotting functions
from src.schedulers.pipedream_1f1b import print_schedule, plot_schedule

# IMPORT THE NEW SCHEDULER (adjust the import path to wherever you saved it)
from src.schedulers.independent_local_sgd_pipeline import IndependentLocalSGDScheduler
```

```python
# problem setup
num_stages = 4
delta_gpd = num_stages**2 - num_stages // 2  # from the theorem I proved

num_parameters = 50

batch_size = 10
total_batches = 30
num_epochs = 1
seed = 0

num_examples = batch_size * total_batches

# In this context, num_microbatches represents the TOTAL passes (M runs * K iterations)
# Here, 10 * 30 = 300 total passes.
num_microbatches = total_batches * num_epochs

objective = QuadraticObjective.synthetic(
    num_examples=num_examples,
    num_parameters=num_parameters,
    num_stages=num_stages,
    batch_size=batch_size,
    seed=seed,
    noise_std=0.0,
)
```

```python
M = 1   # parallel runs
K = 3   # steps per run before syncing
num_rounds = 3
num_microbatches = M * K * num_rounds  # = 200

scheduler = IndependentLocalSGDScheduler(num_runs=M, local_steps=K)
timeline = scheduler.generate(num_stages=num_stages, num_microbatches=num_microbatches)
```

```python
# The existing plot_schedule handles the output natively!
fig, ax = plot_schedule(timeline, startup_boundary=None, reduce_text=True, max_xtick_labels=100)
ax.set_title(f'Independent Local SGD Schedule (M={M} runs)')
plt.show()
```
